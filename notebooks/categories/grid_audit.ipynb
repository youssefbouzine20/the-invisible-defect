# MVTec AD Grid Dataset Audit

Explore the file inventory, integrity checks, and ground-truth masks for the grid category. The tables and figures use the saved measurements from the full dataset audit.

[MVTec AD dataset](https://www.mvtec.com/research-teaching/datasets/mvtec-ad)

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
from IPython.display import display

working_dir = Path.cwd().resolve()
project_root = next(
    (
        folder
        for folder in (working_dir, *working_dir.parents)
        if (folder / "data_tools" / "audit_analysis.py").is_file()
    ),
    None,
)
if project_root is None:
    raise FileNotFoundError("Open the project folder in VS Code and run the notebook again.")

project_path = str(project_root)
if project_path not in sys.path:
    sys.path.insert(0, project_path)

from data_tools.audit_analysis import (
    defect_size_summary,
    integrity_table,
    inventory_table,
    load_category_audit,
    plot_defect_sizes,
    plot_mask_overlays,
)

category = "grid"
dataset_root, audit_report = load_category_audit(project_root, category)
measurements = pd.DataFrame(audit_report["mask_measurements"])
```

```python
print(f"Grid audit: {audit_report['status']}")
display(inventory_table(audit_report))
display(integrity_table(audit_report))
```

## Ground-truth mask overlays

Each row shows one example from a different defect type. The original image is on the left and the ground-truth mask is overlaid in red on the right. Each example has a mask area close to the median for its defect type.

```python
figure = plot_mask_overlays(dataset_root, measurements, category)
plt.show()
plt.close(figure)
```

## Defect size distribution

Defect area is the percentage of the original image pixels marked by the ground-truth mask. The boxplot shows the distribution on a log scale, and the table reports the count, minimum, median, and maximum for each defect type.

```python
figure = plot_defect_sizes(measurements, category)
plt.show()
plt.close(figure)
display(defect_size_summary(measurements))
```
